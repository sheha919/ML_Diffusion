```python
import os
import matplotlib  
matplotlib.use('Agg')  # Use a non-GUI backend
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
cat=['C1']
nrep=12
steps=20000
```

```python
for i in cat:
    for j in range(1, nrep+1):
        path=f'C:/shehani/postdoc_work/ML_Diffusion/cal_md/new_cals/{i}_rel/new/'
        msd_df = pd.read_csv(f'{path}convD_{i}_msd_rep{j}.csv')
        plt.figure()
        plt.plot(msd_df['nstep'], msd_df['msd_x'], label="msd_x")
        plt.plot(msd_df['nstep'], msd_df['msd_y'], label="msd_y")
        plt.plot(msd_df['nstep'], msd_df['msd_z'], label="msd_z")
        plt.xlabel('Number of steps (*10 fs)')
        plt.ylabel('MSD')
        plt.title(f'MSD of OH- over time:{i}-relaxed cation sytem: rep{j}')
        plt.savefig(path +f'{i}_msd_rep_{j}.jpg', dpi=200, bbox_inches='tight') 
        plt.legend()
        plt.show()
```
