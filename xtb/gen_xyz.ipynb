```python
natoms=449
nsheet=287
bx=15.198
by= 13.392
bz=15
file='system'
```

```python
atom= []
x=[]
y=[]
z= []
with open(f'{file}.xyz', 'r') as out_file:
    xyz= out_file.readlines()
   # print(xyz)
    for i in range(len(xyz)):
        atom.append(xyz[i].split()[0])
        x.append(float(xyz[i].split()[1]))
        y.append(float(xyz[i].split()[2]))
        z.append(float(xyz[i].split()[3]))
        
```

```python
#generating xyz file with last step
with open(f"xtb.inp", 'w') as final_xyz:
    final_xyz.write("$chrg 0" + "\n")
    final_xyz.write("$spin 0" + "\n")
    final_xyz.write("$coord angs" + "\n")

    #final_xyz.write(str(natoms) + "\n" + "\n")
    for i in range(natoms):
        final_xyz.write(f"{x[i]:.8f}" + "   " + f"{y[i]:.8f}" + "   " + f"{z[i]:.8f}" + "   " + atom[i]  + "\n")
    final_xyz.write("$end" + "\n"+ "\n")
    final_xyz.write("$hess" + "\n")
    for i in range(1,nsheet+1):
        final_xyz.write(f"modify mass: {i}, 2000000000.0" + "\n")
    final_xyz.write("$end" + "\n")

```
